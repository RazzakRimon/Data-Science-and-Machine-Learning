```python
#function classification

class PythonFunctions:

  # 1. simple function
  def simple_function(self):
    print("This is a simple function")

  # 2. function with parameter
  def function_with_args(self, name):
    print(f"Hello, {name}! welcome")

  # 3. function with return value
  def function_with_return(self, num1, num2):
    return num1 + num2

  # 4. function with default perameter
  def function_with_default(self, country="Bangladesh"):
    print(f"I am from, {country}!")

  # 5. function with keyword argument
  def function_with_kwargs(self, **kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

  # 6. Lamda function
  def lambda_function(self):
    square = lambda x: x * x
    print(f"square(5): {square(5)}")

  # 7. recursive function
  def recursive_function(self, n):
    if n == 1:
      return 1
    else:
      return n * self.recursive_function(n - 1)

# making object of class

pf = PythonFunctions()

# function call

pf.simple_function()
pf.function_with_args("Rimon")
print(f"Result: {pf.function_with_return(5, 30)}")
pf.function_with_default()
pf.function_with_default("India")
pf.function_with_kwargs(name="Rimon", age=25, city="Dhaka")
pf.lambda_function()
print(f"Factorial of 5: {pf.recursive_function(5)}")
```

```python
ages = [5, 12, 17, 18, 24, 32]

def myFunc(x):
 if x < 18:
  return False
 else:
  return True

adults = list(filter(myFunc, ages))

print(adults)
```
