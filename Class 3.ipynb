```python
#Sum of n Numbers Program

n = int(input("Enter the n Number:"))
sum = 0
i = 1

while i <= n:
 sum = sum + i
 i = i +1

print(sum)
```

```python
# For Loop

#Looping Through a String

for x in "banana":
 print(x)
```

```python
#Looping Through a list

fruits = ["apple", "banana", "cherry"]
for x in fruits:
 print(x)
```

```python
#The break Statement
#Exit the loop when x is "banana"

fruits = ["apple", "banana", "cherry"]
for x in fruits:
 print(x)
 if x == "banana":
  break
```

```python
#The continue Statement
#Do not print banana:

fruits = ["apple", "banana", "cherry"]
for x in fruits:
 if x == "banana":
  continue
 print(x)
```

```python
#For with Range Function

#Using the range() function:

for x in range(6):
 print(x)
```

```python
#Increment the sequence with 3 (default is 1):

for x in range(2, 30, 3):
 print(x)
```

```python
#Nested Loops

adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
 for y in fruits:
  print(x, y)
```
